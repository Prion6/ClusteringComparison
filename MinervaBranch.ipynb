```python
import pickle
import os
import gc
import matplotlib.pyplot as plt
import clustering_metrics as metrics
import cluster_utils as cu
import plot_maker as pm
import numpy as np
import seaborn as sns


def check_coincidence(original, variations):

    X_data = original["RA"]
    Y_data = original["DEC"]
    true_labels = original["labels"]

    for idx, variation in variations:

        pred_labels = variation["labels"]
        pm.plot_comparative_clusters(X_data, Y_data, true_labels, pred_labels)
        plt.show()

        data = variation[["RA", "DEC"]].to_numpy(dtype=np.float64)

        completeness, purity, f1 = cu.calculate_overlap_metrics(data, true_labels, pred_labels)

        sns.heatmap(completeness, annot=True, fmt=".2f", cmap="Blues")
        plt.xlabel("Predicted groups")
        plt.ylabel("True sub-halos")
        plt.title("Completeness")
        plt.show()

        sns.heatmap(purity, annot=True, fmt=".2f", cmap="Blues")
        plt.xlabel("Predicted groups")
        plt.ylabel("True sub-halos")
        plt.title("Purity")
        plt.show()

        sns.heatmap(f1, annot=True, fmt=".2f", cmap="Blues")
        plt.xlabel("Predicted groups")
        plt.ylabel("True sub-halos")
        plt.title("F1 Score")
        plt.show()  
```
