```python
import pandas as pd
import glob
import os

# Load all sheet names
folder = "Output/clean_ClusteringComparison"

# Find all .xlsx files in the folder
files = glob.glob(os.path.join(folder, "*.xlsx"))

dfs_nested = []  # array of array of dfs

for f in files:
    xls = pd.ExcelFile(f)
    sheet_names = xls.sheet_names[1:]  # skip the first sheet
    
    file_dfs = []
    for name in sheet_names:
        df = pd.read_excel(xls, sheet_name=name)
        # optional: add metadata if you want to know origin
        df["source_file"] = os.path.basename(f)
        df["sheet_name"] = name
        file_dfs.append(df)
    
    dfs_nested.append(file_dfs)  # add the list of dfs for this file
```
